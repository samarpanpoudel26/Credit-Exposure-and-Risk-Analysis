```python
import pandas as pd
```

```python
#loading data
dataset=pd.read_csv("Data.csv")
dataset.head()
```

```python
#checking null value
dataset.isnull().sum()
```

```python
#Stat
dataset.describe()
```

```python
dataset["card_type"].value_counts()
```

```python
#lower casing the value from columns
dataset["has_chip"]=dataset["has_chip"].str.lower()
dataset["card_on_dark_web"]=dataset["card_on_dark_web"].str.lower()
dataset.head()
```

## Feature Engineering

```python
dataset.head()
```

```python
#Adding column to find out how long ago the pin was changed
dataset["pin_age"]=2026-dataset["year_pin_last_changed"]
dataset.head()
```

```python
#deleting the confidential column which cannot be used during data analysis
dataset=dataset.drop(columns=["card_number","cvv"])
dataset.head()
```

```python
dataset["expires"]=pd.to_datetime(dataset["expires"])
```

```python
dataset['expire_year'] = dataset['expires'].dt.strftime('%Y')
dataset.head()
```

```python
#finding account age 
dataset["expire_year"]=dataset["expire_year"].astype(int)
dataset["account_age"]=2026-dataset["expire_year"]
dataset.head()
```

```python
dataset=dataset.drop(columns=["client_id","expires"])
dataset.head()
```

```python
#creating a seperate column for a multi card client
dataset["multi_card_client"]=["yes" if x>1 else "no" for x in dataset.num_cards_issued]
dataset.head()
```

### Managing credit limit column

```python
dataset['credit_limit'] = dataset['credit_limit'].str.replace('$', '').str.replace(',', '').astype(int)
dataset.head()
```

```python
dataset.credit_limit.describe()
```

```python
#Adding a column for a types of credit limit
conditions= [
    dataset["credit_limit"]<=1000,
    (dataset["credit_limit"]>1000) & (dataset["credit_limit"]<=5000),
    (dataset["credit_limit"]>5000) & (dataset["credit_limit"]<10000),
    dataset["credit_limit"]>=10000
]
choices=["low","medium","high","premium"]

import numpy as np
dataset['limit_category'] = np.select(conditions, choices, default='F')
dataset.head()
```

```python
dataset["limit_category"].unique()
```

## Saving the cleaned file

```python
dataset.to_csv("processed.csv",index=False)
```
